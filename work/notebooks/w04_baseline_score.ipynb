## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
!git clone https://github.com/Vaanya0730/flyrank-ml-internship-starter.git /content/flyrank-ml-internship-starter
import pandas as pd

df = pd.read_csv("/content/flyrank-ml-internship-starter/data/raw/content_refresh_anonymized.csv")
df.columns
import pandas as pd
import numpy as np

df = pd.read_csv("/content/flyrank-ml-internship-starter/data/raw/content_refresh_anonymized.csv")

# Replace avg_position=0 (means no data)
df["avg_position"] = df["avg_position"].replace(0, np.nan)

# Normalize features
df["imp_score"] = df["impressions_90d"] / df["impressions_90d"].max()

# Lower CTR = bigger opportunity
df["ctr_score"] = 1 - (df["ctr"] / df["ctr"].max())

# Best opportunity around positions 8-20
df["position_score"] = np.where(
    (df["avg_position"] >= 8) &
    (df["avg_position"] <= 20),
    1,
    0
)

df[["imp_score","ctr_score","position_score"]].head()
```


## Rule

I rank content using a simple rule based on search opportunity.

Pages receive a higher score when they have:
- High impressions
- Low CTR
- Average search position between 8 and 20

These pages already receive visibility but are not getting enough clicks, so improving titles or meta descriptions could increase traffic.

## Reason Codes

RC01 – High impressions with low CTR

Action:
Improve title and meta description to increase CTR.

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Weighted baseline score
df["baseline_action_score"] = (
    0.5 * df["imp_score"] +
    0.3 * df["ctr_score"] +
    0.2 * df["position_score"]
)

df["reason_code"] = "RC01"
df["action"] = "Improve title/meta description"

ranked = df.sort_values(
    "baseline_action_score",
    ascending=False
)

import os

output_dir = "/content/flyrank-ml-internship-starter/work/outputs"
os.makedirs(output_dir, exist_ok=True)

ranked.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

print("CSV written successfully.")
ranked.head(20)
```

## Ranked Queue

The score combines three signals.

- High impressions
- Low CTR
- Position between 8 and 20

Higher scores indicate pages with greater optimization opportunity.
