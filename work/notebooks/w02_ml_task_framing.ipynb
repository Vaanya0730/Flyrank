## ML Task Type: Scoring

This lane is best framed as a scoring problem. Instead of assigning a simple yes/no label, the model would generate an opportunity score for each content page. Pages with higher scores would be prioritized for refresh because they are more likely to benefit from updates based on search performance and engagement signals.
